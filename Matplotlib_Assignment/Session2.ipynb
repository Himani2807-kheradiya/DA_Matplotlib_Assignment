TASK:1.
Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.

```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6000, 7500, 5000, 8000, 10000, 8500]
```

```python
plt.plot(days, steps, color="green", linewidth=2, marker="o")

plt.title("Daily Steps Walked Over 7 Days")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Steps")

plt.grid(True)
plt.show()
```

TASK:3.
Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.

```python
months = ["May", "Jun", "Jul", "Aug", "Sep"]
```

```python
zomato_orders = [8, 10, 7, 12, 15]
swiggy_orders = [6, 9, 11, 8, 13]
```

```python
plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)
```

```python
plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="s",
    label="Swiggy"
)
```

```python
plt.title("Monthly Orders: Zomato vs Swiggy")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.legend()
plt.grid(True)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)


plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="s",
    label="Swiggy"
)

plt.title("Monthly Orders: Zomato vs Swiggy")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()
```
