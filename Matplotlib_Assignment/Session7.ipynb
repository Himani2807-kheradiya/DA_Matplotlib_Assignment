```python
plt.subplot(2, 1, 2)
plt.plot(days, water_intake, color="green", marker="o")
plt.title("Daily Water Intake")
plt.xlabel("Day")
plt.ylabel("Water Intake (Litres)")
plt.grid(True, linestyle="--", alpha=0.5)
```

TASK:2.
Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

```python
quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [25, 40, 35, 50]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(quarters, posts, color="blue", marker="o")
ax[0, 0].set_title("Line Chart - Quarterly Posts")
ax[0, 0].set_xlabel("Quarter")
ax[0, 0].set_ylabel("Number of Posts")
ax[0, 0].grid(True, linestyle="--", alpha=0.5)

ax[0, 1].bar(quarters, posts, color="orange")
ax[0, 1].set_title("Bar Chart - Quarterly Posts")
ax[0, 1].set_xlabel("Quarter")
ax[0, 1].set_ylabel("Number of Posts")

ax[1, 0].scatter(quarters, posts, color="green", s=100)
ax[1, 0].set_title("Scatter Plot - Quarterly Posts")
ax[1, 0].set_xlabel("Quarter")
ax[1, 0].set_ylabel("Number of Posts")
ax[1, 0].grid(True, linestyle="--", alpha=0.5)

ax[1, 1].pie(
    posts,
    labels=quarters,
    autopct="%1.1f%%",
    startangle=90
)
ax[1, 1].set_title("Pie Chart - Share of Annual Posts")

plt.tight_layout()
plt.show()
```

TASK:3.
Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

```python
months = ["May", "Jun", "Jul", "Aug", "Sep", "Oct"]
```

```python
zomato_orders = [8, 10, 7, 12, 9, 11]
swiggy_orders = [6, 9, 8, 10, 7, 12]
dominos_orders = [4, 5, 6, 3, 7, 5]
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

ax[0].plot(months, zomato_orders, marker="o", color="red")
ax[0].set_title("Zomato Orders")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Number of Orders")
ax[0].grid(True, linestyle="--", alpha=0.5)


ax[1].plot(months, swiggy_orders, marker="o", color="orange")
ax[1].set_title("Swiggy Orders")
ax[1].set_xlabel("Month")
ax[1].grid(True, linestyle="--", alpha=0.5)

ax[2].plot(months, dominos_orders, marker="o", color="blue")
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel("Month")
ax[2].grid(True, linestyle="--", alpha=0.5)

fig.suptitle("Food Orders Over the Last 6 Months", fontsize=14)

plt.tight_layout()
plt.show()
```

TASK:4.
Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: YouTube, Instagram, Spotify, and WhatsApp. Ensure all subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>

```python
months = ["May", "Jun", "Jul", "Aug", "Sep", "Oct"]
```

```python
youtube = [8.5, 10.2, 12.0, 9.8, 13.5, 15.0]
instagram = [3.2, 4.0, 3.8, 4.5, 5.0, 4.7]
spotify = [1.5, 1.8, 2.0, 1.7, 2.2, 2.5]
whatsapp = [0.8, 1.0, 0.9, 1.2, 1.1, 1.3]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8), sharex=True)


ax[0, 0].plot(months, youtube, marker="o", color="red")
ax[0, 0].set_title("YouTube Data Usage")
ax[0, 0].set_ylabel("Data Usage (GB)")
ax[0, 0].grid(True, linestyle="--", alpha=0.5)


ax[0, 1].plot(months, instagram, marker="o", color="purple")
ax[0, 1].set_title("Instagram Data Usage")
ax[0, 1].set_ylabel("Data Usage (GB)")
ax[0, 1].grid(True, linestyle="--", alpha=0.5)


ax[1, 0].plot(months, spotify, marker="o", color="green")
ax[1, 0].set_title("Spotify Data Usage")
ax[1, 0].set_xlabel("Month")
ax[1, 0].set_ylabel("Data Usage (GB)")
ax[1, 0].grid(True, linestyle="--", alpha=0.5)

ax[1, 1].plot(months, whatsapp, marker="o", color="blue")
ax[1, 1].set_title("WhatsApp Data Usage")
ax[1, 1].set_xlabel("Month")
ax[1, 1].set_ylabel("Data Usage (GB)")
ax[1, 1].grid(True, linestyle="--", alpha=0.5)

fig.suptitle("Monthly Mobile Data Usage by App", fontsize=15)


plt.tight_layout()
plt.show()
```
