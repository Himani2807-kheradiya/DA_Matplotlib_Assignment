TASK:1.
Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

```python
import matplotlib.pyplot as plt
```

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']
```

```python
zomato_orders = [120, 150, 135, 180, 200, 250, 230]
swiggy_orders = [100, 140, 160, 170, 190, 220, 260]
```

```python
plt.bar(days, zomato_orders, color='red', label='Zomato')
plt.bar(days, swiggy_orders, bottom=zomato_orders,
        color='orange', label='Swiggy')

plt.title('Daily Food Orders: Zomato vs Swiggy')
plt.xlabel('Days of the Week')
plt.ylabel('Number of Orders')

plt.legend()

plt.show()
```

TASK:3.
Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>


```python
days = list(range(1, 11))

```

```python
avg_delivery_time = [35, 40, 38, 45, 42, 50, 44, 39, 47, 43]
```

```python
std_deviation = [3, 4, 2, 5, 3, 6, 4, 2, 5, 3]
```

```python
plt.errorbar(
    days,
    avg_delivery_time,
    yerr=std_deviation,
    fmt='o-',
    color='blue',
    ecolor='red',
    capsize=5,
    label='Average Delivery Time'
)

plt.title('Flipkart Daily Average Delivery Time')
plt.xlabel('Day')
plt.ylabel('Delivery Time (minutes)')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

TASK:4.
Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

```python
import pandas as pd
import matplotlib.dates as mdates
```

```python
dates = pd.date_range(end=pd.Timestamp.today(), periods=14)
```

```python
balance = [2500, 2300, 2800, 2200, 2600, 2400, 3000,
           2700, 3200, 2900, 3100, 2800, 3500, 3300]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(dates, balance, marker='o', color='blue')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))


plt.title('Paytm Wallet Balance - Last 14 Days')
plt.xlabel('Date')
plt.ylabel('Wallet Balance (₹)')

plt.xticks(rotation=45)

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Refactor your area chart code to add gridlines and custom axis titles, making the chart look more like the analytics dashboards you see in apps like Instagram Insights.<br><br><em><strong>Constraint:</strong> Use ax.set_title(), ax.set_xlabel(), ax.set_ylabel(), and ax.grid(True) in your code.</em>

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.fill_between(days[:len(spotify_users)], spotify_users, color='green',
                alpha=0.4, label='Spotify')
ax.fill_between(days[:len(youtube_users)], youtube_users, color='purple',
                 alpha=0.4, label='YouTube')

ax.set_title('Weekly Active Users - Analytics Dashboard')
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Daily Active Users')

ax.grid(True)

ax.legend()

plt.tight_layout()
plt.show()
```
