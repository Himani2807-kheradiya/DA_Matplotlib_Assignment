TASK:1.
Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.

```python
import matplotlib.pyplot as plt
```

```python
restaurants = ['Dominos', 'Pizza Hut', 'McDonalds',
               'Burger King', 'Subway']
orders = [150, 120, 180, 100, 90]
```

```python
plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders, color='orange')

plt.title('Number of Orders for Zomato Restaurants')
plt.xlabel('Restaurants')
plt.ylabel('Number of Orders')

plt.xticks(rotation=20)

plt.savefig('zomato_orders.png', dpi=150, bbox_inches='tight')

plt.show()
```

TASK:3.
Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

```python
apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify']
time_spent = [3, 4, 2, 1]
```

```python
plt.figure(figsize=(7, 7))
plt.pie(
    time_spent,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90,
    colors=["pink", 'red', 'green', 'brown']
)

plt.title('Daily Time Spent on Apps')

plt.savefig('daily_app_usage.pdf', format='pdf', bbox_inches='tight')

plt.show()
```

TASK:4.
Take any plot you created above and embed the saved image into a PowerPoint slide using PowerPoint (or Google Slides).<br><br><em><strong>Hint:</strong> Use the Insert > Picture option in PPT to add your exported chart.</em>

```python
%pip install matplotlib python-pptx
```

```python
import matplotlib.pyplot as plt
from pptx import Presentation
from pptx.util import Inches

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

steps = [4500, 6000, 5200, 7500, 8000, 10000, 9000]

plt.figure(figsize=(8, 4))
plt.plot(days, steps, marker='o', color='blue')

plt.title('Daily Step Count for One Week')
plt.xlabel('Days')
plt.ylabel('Number of Steps')
plt.grid(True)

plt.savefig('daily_steps.png', dpi=150, bbox_inches='tight')
plt.close()

prs = Presentation()

slide = prs.slides.add_slide(prs.slide_layouts[5])
slide.shapes.title.text = 'Daily Step Count for One Week'

slide.shapes.add_picture(
    'daily_steps.png',
    Inches(1),
    Inches(1.5),
    width=Inches(11)
)

prs.save('Daily_Step_Count.pptx')

print("Chart inserted into PowerPoint successfully!")
print("Saved as Daily_Step_Count.pptx")
```

TASK:5.
Export a Matplotlib chart as a PDF, then verify the PDF opens correctly and the chart looks sharp (not blurry) at 200% zoom. If the image looks pixelated, try exporting again with a higher DPI setting and compare the results.

```python

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6000, 5200, 7500, 8000, 10000, 9000]

plt.figure(figsize=(9, 5.5))
plt.plot(days, steps, marker="o", linewidth=2, markersize=7)

plt.title("Daily Step Count for One Week")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig("daily_steps_300dpi.pdf", dpi=300, format="pdf")

plt.savefig("daily_steps_600dpi.pdf", dpi=600, format="pdf")

plt.show()
```
