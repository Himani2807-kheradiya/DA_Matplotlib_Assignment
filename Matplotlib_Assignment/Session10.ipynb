TASK:1.
Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

```python
active_users = [12000, 13500, 12800, 15000, 16500, 18000,
                17500, 19000, 21000, 20500, 23000, 25000]
```

```python
max_users = max(active_users)
max_index = active_users.index(max_users)
max_month = months[max_index]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(months, active_users, marker='o', color='green',
         label='Daily Active Users')

plt.annotate(
    f'Highest: {max_month} ({max_users:,} users)',
    xy=(max_month, max_users),
    xytext=(-100, -40),
    textcoords='offset points',
    arrowprops=dict(arrowstyle='->', color='red'),
    color='red',
    fontsize=10
)

plt.title('Monthly Daily Active Users - Music Streaming App')
plt.xlabel('Month')
plt.ylabel('Daily Active Users')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()
```

TASK:2.
Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

```python
orders = [1200, 1500, 1100, 1800, 2000, 1700,
          2200, 2500, 1900, 1600, 2300, 2100]
```

```python
min_orders = min(orders)
min_index = orders.index(min_orders)
min_month = months[min_index]
```

```python
plt.figure(figsize=(10, 5))
plt.bar(months, orders, color='orange', edgecolor='black')

plt.annotate(
    f'Lowest: {min_month} ({min_orders} orders)',
    xy=(min_index, min_orders),
    xytext=(min_index + 1, min_orders + 500),
    arrowprops=dict(arrowstyle='->', color='green', lw=2),
    color='green',
    fontsize=10
)

plt.title('Monthly Orders - Zomato')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

TASK:3.
Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

```python
payment_methods = ['UPI', 'Card', 'Wallet', 'Cash']
percentages = [45, 30, 15, 10]
```

```python
plt.figure(figsize=(7, 7))
plt.pie(
    percentages,
    labels=payment_methods,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Payment Methods Used on Flipkart')

plt.legend(payment_methods, loc='lower left')

plt.show()
```

TASK:4.
Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

```python
movies = ['Movie A', 'Movie B', 'Movie C',
          'Movie D', 'Movie E']
```

```python
ratings = [7.5, 8.2, 9.1, 6.8, 8.7]
reviews = [1200, 2500, 1800, 900, 3000]
```

```python
max_index = ratings.index(max(ratings))
best_movie = movies[max_index]
best_rating = ratings[max_index]
best_reviews = reviews[max_index]
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(reviews, ratings, color='blue', s=100)

plt.annotate(
    f'{best_movie}\nRating: {best_rating}',
    xy=(best_reviews, best_rating),
    xytext=(best_reviews + 200, best_rating - 0.5),
    bbox=dict(boxstyle='round,pad=0.5',
              fc='yellow', alpha=0.8),
    arrowprops=dict(arrowstyle='->', color='red')
)

plt.title('Movie Ratings vs. Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Movie Rating')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
