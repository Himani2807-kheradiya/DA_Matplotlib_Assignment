TASK:1.
Create a line chart showing the number of daily active users on a music streaming app (like Spotify) over 7 days, and set the figure size to 10 inches wide by 4 inches tall using plt.figure().

```python
import matplotlib.pyplot as plt
```

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']
```

```python
active_users = [1200, 1500, 1350, 1800, 2100, 2500, 2300]
```

```python
plt.figure(figsize=(10, 4))
```

```python
plt.plot(days, active_users, marker='o', color='green')

plt.title('Daily Active Users - Music Streaming App')
plt.xlabel('Days')
plt.ylabel('Number of Active Users')

plt.grid(True)

plt.show()
```

TASK:2.
Plot two lines on the same graph representing the number of food orders on Zomato and Swiggy over a week, using different line styles (e.g., dashed for Zomato, dotted for Swiggy) and add a legend to distinguish them.


```python
zomato_orders = [120, 150, 135, 180, 200, 250, 230]
swiggy_orders = [100, 140, 160, 170, 190, 220, 260]
```

```python
plt.plot(days, zomato_orders, linestyle='--', marker='o',
         label='Zomato')

plt.plot(days, swiggy_orders, linestyle=':', marker='s',
         label='Swiggy')

plt.title('Weekly Food Orders: Zomato vs Swiggy')
plt.xlabel('Days of the Week')
plt.ylabel('Number of Orders')

plt.legend()
plt.grid(True)

plt.show()
```

TASK:3.
Apply the 'ggplot' style to a chart showing Flipkart's daily sales for 10 days using plt.style.use('ggplot'), and customize the line color to orange.


```python
plt.style.use('ggplot')
```

```python
days = list(range(1, 11))
daily_sales = [12000, 15000, 13500, 18000, 20000,
               17500, 22000, 25000, 23000, 28000]
```

```python
plt.plot(days, daily_sales, color='orange', marker='o')

plt.title("Flipkart Daily Sales for 10 Days")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show()
```
