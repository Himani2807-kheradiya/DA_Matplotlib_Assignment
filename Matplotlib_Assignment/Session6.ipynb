TASK:2.
Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

```python
friends = [
    "Amit", "Priya", "Rahul", "Neha",
    "Karan", "Sneha", "Vikas", "Pooja",
    "Rohan", "Anjali", "Arjun", "Meera"
]
```

```python
amount_spent = [1200, 2500, 1800, 3200, 900, 2100,
                1500, 2800, 3500, 1100, 2400, 1900]

number_of_orders = [4, 8, 6, 10, 3, 7,
                    5, 9, 11, 4, 8, 6]
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(number_of_orders, amount_spent, color="red", marker="o")

for i in range(len(friends)):
    plt.annotate(
        friends[i],
        (number_of_orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Zomato Spending vs Number of Orders")
plt.xlabel("Number of Orders Placed")
plt.ylabel("Amount Spent on Zomato (₹)")

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

TASK:3.Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
friends = [
    "Amit", "Priya", "Rahul", "Neha",
    "Karan", "Sneha", "Vikas", "Pooja",
    "Rohan", "Anjali", "Arjun", "Meera"
]
```

```python
amount_spent = [1200, 2500, 1800, 3200, 900, 2100,
                1500, 2800, 3500, 1100, 2400, 1900]

number_of_orders = [4, 8, 6, 10, 3, 7,
                    5, 9, 11, 4, 8, 6]
```

```python
slope, intercept = np.polyfit(number_of_orders, amount_spent, 1)
```

```python
trend_line = np.poly1d([slope, intercept])
x_line = np.linspace(min(number_of_orders), max(number_of_orders), 100)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(number_of_orders, amount_spent,
            color="red", label="Friends' Orders")

for i in range(len(friends)):
    plt.annotate(
        friends[i],
        (number_of_orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.plot(x_line, trend_line(x_line),
         color="blue", linestyle="--",
         linewidth=2, label="Best-Fit Trend Line")

plt.title("Zomato Spending vs Number of Orders")
plt.xlabel("Number of Orders Placed")
plt.ylabel("Amount Spent on Zomato (₹)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print("Slope:", slope)
print("Intercept:", intercept)
```

TASK:4.
Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

```python
youtube_hours = [2, 4, 3, 5, 6, 1, 7, 4, 8, 5, 9, 3, 6, 10, 7]
spotify_hours = [3, 5, 4, 6, 7, 2, 8, 5, 9, 6, 10, 4, 7, 11, 8]
```

```python
slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)
```

```python
x_line = np.linspace(min(youtube_hours), max(youtube_hours), 100)
y_line = slope * x_line + intercept
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(youtube_hours, spotify_hours,
            color="purple", label="Users")
```

```python
plt.plot(x_line, y_line, color="red",
         linestyle="--", label="Regression Line")
```

```python
equation = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(
    0.05, 0.95, equation,
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment="top",
    bbox=dict(facecolor="white", edgecolor="black")
)
```

```python
plt.title("YouTube Watch Hours vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print("Regression equation:", equation)
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(youtube_hours, spotify_hours,
            color="purple", label="Users")

plt.plot(x_line, y_line, color="red",
         linestyle="--", label="Regression Line")

equation = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(
    0.05, 0.95, equation,
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment="top",
    bbox=dict(facecolor="white", edgecolor="black")
)

plt.title("YouTube Watch Hours vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print("Regression equation:", equation)
```
