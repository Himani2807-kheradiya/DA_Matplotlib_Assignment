TASK:2.
Using Matplotlib, create a bar chart that shows total sales for each product category (like 'Electronics', 'Clothing', 'Groceries'), similar to how Flipkart or Myntra might track category performance.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python


df = pd.read_csv("sample_sales.csv")
```

```python
category_sales = df.groupby("category")["sales_amount"].sum()
```

```python
plt.bar(category_sales.index, category_sales.values,
        color="skyblue", edgecolor="black")

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")

for i, value in enumerate(category_sales.values):
    plt.text(i, value, str(value), ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

TASK:3.
Plot a line chart to visualize monthly sales trends over a year, using the date column to group sales by month. Add a title and axis labels to make the chart presentation-ready.

```python
data = {
    "date": [
        "2026-01-10", "2026-02-10", "2026-03-10",
        "2026-04-10", "2026-05-10", "2026-06-10",
        "2026-07-10", "2026-08-10", "2026-09-10",
        "2026-10-10", "2026-11-10", "2026-12-10"
    ],
    "sales_amount": [
        12000, 15000, 18000, 14000, 22000, 25000,
        20000, 28000, 30000, 26000, 35000, 40000
    ]
}
```

```python
df = pd.DataFrame(data)
df.to_csv("monthly_sales.csv", index=False)
```

```python
df = pd.read_csv("monthly_sales.csv")
df["date"] = pd.to_datetime(df["date"])

monthly_sales = df.groupby(df["date"].dt.to_period("M"))["sales_amount"].sum()

plt.figure(figsize=(10, 5))
plt.plot(
    range(1, len(monthly_sales) + 1),
    monthly_sales.values,
    marker="o",
    linestyle="-",
    color="blue"
)

plt.title("Monthly Sales Trends for 2026")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(range(1, 13))
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

TASK:5.
Add value annotations to your category bar chart to display the exact sales amount above each bar, then export the chart as a PNG image file named 'category_sales_kpi.png'.

```python
df = pd.read_csv("sample_sales.csv")
```

```python
category_sales = df.groupby("category")["sales_amount"].sum()
```

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(
    category_sales.index,
    category_sales.values,
    color="skyblue",
    edgecolor="black"
)
```

```python



plt.figure(figsize=(8, 5))
bars = plt.bar(
    category_sales.index,
    category_sales.values,
    color="skyblue",
    edgecolor="black"
)

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=11
    )

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")

plt.tight_layout()

plt.savefig("category_sales_kpi.png", dpi=300, bbox_inches="tight")

plt.show()
```
