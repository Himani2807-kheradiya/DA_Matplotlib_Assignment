TASK:1.
Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders: [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42]. Show the plot window.

```python
import matplotlib.pyplot as plt
```

```python
delivery_times = [
    32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
    27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
    36, 28, 43, 31, 35, 40, 38, 33, 30, 42
]
```

```python
plt.hist(delivery_times, bins=6, color="orange", edgecolor="black")


plt.title("Swiggy Order Delivery Time Distribution")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")


plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

TASK:2.
Plot a histogram of IPL match scores (runs per team in a match) using plt.hist() and experiment with bins=5, bins=10, and bins=15. Observe and describe how changing the number of bins affects the shape of the histogram.<br><br><em><strong>Hint:</strong> Use any sample array of 50-60 scores between 120 and 220.</em>

```python
scores = [
    145, 167, 182, 195, 156, 210, 178, 165, 190, 201,
    134, 172, 188, 220, 159, 181, 197, 148, 205, 169,
    175, 193, 160, 214, 184, 152, 199, 177, 186, 142,
    208, 163, 191, 155, 218, 170, 180, 196, 147, 203,
    166, 187, 212, 158, 194, 173, 185, 139, 200, 179,
    162, 206, 150, 189, 216, 171
]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Histogram with 5 bins
axes[0].hist(scores, bins=5, color="blue", edgecolor="black")
axes[0].set_title("IPL Scores (5 Bins)")
axes[0].set_xlabel("Runs")
axes[0].set_ylabel("Frequency")

axes[1].hist(scores, bins=10, color="green", edgecolor="black")
axes[1].set_title("IPL Scores (10 Bins)")
axes[1].set_xlabel("Runs")
axes[1].set_ylabel("Frequency")

axes[2].hist(scores, bins=15, color="orange", edgecolor="black")
axes[2].set_title("IPL Scores (15 Bins)")
axes[2].set_xlabel("Runs")
axes[2].set_ylabel("Frequency")


plt.suptitle("Distribution of IPL Team Scores")
plt.tight_layout()
plt.show()
```

TASK:3.
Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.

```python
steps = [
    4500, 6000, 7500, 8000, 5500, 9000, 10000,
    6500, 7000, 8500, 4000, 9500, 11000, 7500,
    6000, 5000, 8000, 9000, 10500, 7000, 8500,
    5500, 6500, 12000, 10000, 7500, 9000, 5000,
    8000, 11000
]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histogram 1: Frequency (default)
axes[0].hist(steps, bins=6, color="blue", edgecolor="black")
axes[0].set_title("Daily Steps - Frequency")
axes[0].set_xlabel("Number of Steps")
axes[0].set_ylabel("Frequency (Number of Days)")
```

```python

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].hist(steps, bins=6, color="blue", edgecolor="black")
axes[0].set_title("Daily Steps - Frequency")
axes[0].set_xlabel("Number of Steps")
axes[0].set_ylabel("Frequency (Number of Days)")
axes[1].hist(
    steps,
    bins=6,
    density=True,
    color="green",
    edgecolor="black"
)
axes[1].set_title("Daily Steps - Density")
axes[1].set_xlabel("Number of Steps")
axes[1].set_ylabel("Probability Density")

plt.tight_layout()
plt.show()
```
