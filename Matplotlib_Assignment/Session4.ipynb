TASK:1.
Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.

```python
import matplotlib.pyplot as plt
```

```python
apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
```

```python
hours = [3, 4, 2, 1, 2]
```

```python
plt.pie(
    hours,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)


plt.title("Daily Time Spent on Different Apps")


plt.show()
```

TASK:3.
Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.

```python
categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]
```

```python
spending = [3000, 5000, 2000, 4000]
```

```python
explode = [0, 0.15, 0, 0]
```

```python
fig, ax = plt.subplots()
```

```python
fig, ax = plt.subplots()

ax.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90
)

circle = plt.Circle((0, 0), 0.60, color="white")
ax.add_artist(circle)

ax.set_title("Monthly Online Spending Donut Chart")

ax.axis("equal")

plt.show()
```

TASK:4.Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>


```python
teams = ["CSK", "MI", "RCB", "KKR"]
followers = [15, 14, 17, 8]
```

```python
fig, ax = plt.subplots()

ax.pie(
    followers,
    labels=teams,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.75
)


circle = plt.Circle((0, 0), 0.50, color="white")
ax.add_artist(circle)

ax.set_title("IPL Teams Instagram Followers Share")


ax.axis("equal")


plt.show()
```
