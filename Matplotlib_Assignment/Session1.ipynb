```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
steps = [4500, 6000, 7500, 5000, 8000, 10000, 8500]
```

```python
plt.plot(days, steps, marker="o", color="red", linestyle="-")
plt.title("Daily Steps Walked Over 7 Days")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Steps")
plt.grid(True)
plt.tight_layout()
plt.show()
```

TASK:3.
Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
spotify_hours = [1, 2, 1.5, 2.5, 3, 2, 1]
youtube_hours = [2, 1.5, 2, 3, 2.5, 4, 3.5]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].plot(days, spotify_hours, marker="o", color="green")
axes[0].set_title("Daily Spotify Listening Hours")
axes[0].set_xlabel("Day of the Week")
axes[0].set_ylabel("Listening Hours")
axes[0].grid(True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[1].plot(days, youtube_hours, marker="o", color="red")
axes[1].set_title("Daily YouTube Viewing Hours")
axes[1].set_xlabel("Day of the Week")
axes[1].set_ylabel("Viewing Hours")
axes[1].grid(True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].plot(days, spotify_hours, marker="o", color="green")
axes[0].set_title("Daily Spotify Listening Hours")
axes[0].set_xlabel("Day of the Week")
axes[0].set_ylabel("Listening Hours")
axes[0].grid(True)
axes[1].plot(days, youtube_hours, marker="o", color="red")
axes[1].set_title("Daily YouTube Viewing Hours")
axes[1].set_xlabel("Day of the Week")
axes[1].set_ylabel("Viewing Hours")
axes[1].grid(True)
plt.tight_layout()
plt.show()
```

TASK:4.
Add axis labels, a title, and a grid to your plot from the previous task to clearly communicate the information, mimicking the style of analytics dashboards in apps like Instagram Insights.<br><br><em><strong>Hint:</strong> Use set_xlabel(), set_ylabel(), set_title(), and grid() methods on the Axes object.</em>

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
spotify_hours = [1, 2, 1.5, 2.5, 3, 2, 1]
youtube_hours = [2, 1.5, 2, 3, 2.5, 4, 3.5]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(days, spotify_hours, marker="o", color="green")
ax1.set_xlabel("Day of the Week")
ax1.set_ylabel("Listening Hours")
ax1.set_title("Weekly Spotify Listening Analytics")
ax1.grid(True, linestyle="--", alpha=0.6)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax2.plot(days, youtube_hours, marker="o", color="red")
ax2.set_xlabel("Day of the Week")
ax2.set_ylabel("Viewing Hours")
ax2.set_title("Weekly YouTube Viewing Analytics")
ax2.grid(True, linestyle="--", alpha=0.6)
fig.suptitle("Weekly Media Usage Dashboard", fontsize=16)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(days, spotify_hours, marker="o", color="green")
ax1.set_xlabel("Day of the Week")
ax1.set_ylabel("Listening Hours")
ax1.set_title("Weekly Spotify Listening Analytics")
ax1.grid(True, linestyle="--", alpha=0.6)

ax2.plot(days, youtube_hours, marker="o", color="red")
ax2.set_xlabel("Day of the Week")
ax2.set_ylabel("Viewing Hours")
ax2.set_title("Weekly YouTube Viewing Analytics")
ax2.grid(True, linestyle="--", alpha=0.6)

fig.suptitle("Weekly Media Usage Dashboard", fontsize=16)

plt.tight_layout()
plt.show()
```
